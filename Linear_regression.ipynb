```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
data = {
    "AdBudget": [5, 8, 10, 12, 15, 18, 20, 23, 25, 28, 30, 33, 35, 38, 40, 43, 45, 48, 50, 55],
    "Sales": [20, 24, 28, 31, 36, 40, 44, 47, 51, 55, 58, 63, 66, 70, 73, 77, 80, 84, 88, 95]
}
df = pd.DataFrame(data)
```

```python
X = df[["AdBudget"]]
y = df["Sales"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=7)
```

```python
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)
```

```python
sales_prediction = lr_model.predict(X_test)
```

```python
mse = mean_squared_error(y_test, sales_prediction)
r2 = r2_score(y_test, sales_prediction)

print("Model Results")
print("-" * 30)
print("Mean Squared Error:", round(mse, 2))
print("R2 Score:", round(r2, 2))
print("Intercept:", round(lr_model.intercept_, 2))
print("Coefficient:", round(lr_model.coef_[0], 2))
```
