```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
data = {
    "AdBudget": [5, 8, 10, 12, 15, 18, 20, 23, 25, 28,
                 30, 33, 35, 38, 40, 43, 45, 48, 50, 55],
    "Sales": [20, 24, 28, 31, 36, 40, 44, 47, 51, 55,
              58, 63, 66, 70, 73, 77, 80, 84, 88, 95]
}
df = pd.DataFrame(data)
X = df[["AdBudget"]]
y = df["Sales"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=7
)
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)
sales_prediction = lr_model.predict(X_test)
mse = mean_squared_error(y_test, sales_prediction)
r2 = r2_score(y_test, sales_prediction)

print("Model Results")
print("-" * 30)
print("Mean Squared Error:", round(mse, 2))
print("R2 Score:", round(r2, 2))
print("Intercept:", round(lr_model.intercept_, 2))
print("Coefficient:", round(lr_model.coef_[0], 2))
sorted_index = X_test["AdBudget"].argsort()
X_sorted = X_test.iloc[sorted_index]
y_sorted = sales_prediction[sorted_index]
plt.figure(figsize=(9, 5))

plt.scatter(
    X_test,
    y_test,
    color="teal",
    marker="D",
    s=80,
    label="Actual Sales"
)

plt.plot(
    X_sorted,
    y_sorted,
    color="crimson",
    linewidth=3,
    label="Best Fit Line"
)

plt.title("Advertising Budget vs Product Sales")
plt.xlabel("Advertising Budget (Thousands of Dollars)")
plt.ylabel("Product Sales (Thousands of Units)")
plt.grid(alpha=0.4)
plt.legend()

plt.show()
```
